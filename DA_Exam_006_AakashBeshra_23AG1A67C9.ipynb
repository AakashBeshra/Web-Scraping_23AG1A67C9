```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import re
from sklearn.preprocessing import MinMaxScaler
data = pd.read_csv("/content/Car_Price_Performance.csv")
brand_avg_prices = data.groupby('Brand')['Price'].mean().sort_values()
brand_mapping = {brand: i for i, brand in enumerate(brand_avg_prices.index)}
data['Brand_encoded'] = data['Brand'].map(brand_mapping)
scaler = MinMaxScaler()
data[['Horsepower', 'Fuel Efficiency']] = scaler.fit_transform(data[['Horsepower', 'Fuel Efficiency']])
data['Scaled_Brand_encoded'] = scaler.fit_transform(data[['Brand_encoded']])
data['Performance_Index'] = (data['Horsepower'] * 0.7 + data['Fuel Efficiency'] * 0.3) / data['Price']
result_table = data[['Brand', 'Price', 'Horsepower', 'Fuel Efficiency', 'Brand_encoded', 'Horsepower', 'Fuel Efficiency', 'Scaled_Brand_encoded', 'Performance_Index']]
result_table = result_table.rename(columns={
    'Horsepower': 'Scaled_Horsepower',
    'Fuel Efficiency': 'Scaled_Fuel_Efficiency',
    'Scaled_Brand_encoded': 'Scaled_Brand'
})
print(result_table.head())
```
